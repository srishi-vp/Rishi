```python
# Step 1: Import spaCy
import spacy

# Step 2: Load English language model
nlp = spacy.load("en_core_web_sm")

# Step 3: Input text
text = "Apple is looking at buying a startup in Bengaluru for $1 billion."

# Step 4: Process the text
doc = nlp(text)

# -----------------------------
# Tokenization
# -----------------------------
print("Tokenization:")
for token in doc:
    print(token.text)

# -----------------------------
# Stemming (SpaCy does not support stemming directly)
# Using lemma as an alternative
# -----------------------------
print("\nStemming (using lemma as approximation):")
for token in doc:
    print(token.text, "->", token.lemma_)

# -----------------------------
# Lemmatization
# -----------------------------
print("\nLemmatization:")
for token in doc:
    print(token.text, "->", token.lemma_)

# -----------------------------
# POS Tagging
# -----------------------------
print("\nPOS Tagging:")
for token in doc:
    print(token.text, "->", token.pos_)

# -----------------------------
# Named Entity Recognition (NER)
# -----------------------------
print("\nNamed Entity Recognition:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)

# -----------------------------
# Stop Word Removal
# -----------------------------
print("\nStop Word Removal:")
filtered_tokens = [token.text for token in doc if not token.is_stop]
print(filtered_tokens)
```
