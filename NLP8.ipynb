```python
# EXP NO: 8
# Text Modeling using TF-IDF and N-gram Model

# Step 1: Import required libraries
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer

# Step 2: Sample text documents
documents = [
    "Natural language processing is a part of artificial intelligence",
    "Machine learning helps computers understand language",
    "Text mining and NLP are used for data analysis",
    "Artificial intelligence and machine learning are related fields"
]

print("Documents:")
for i, doc in enumerate(documents):
    print(i+1, ":", doc)


# -----------------------------------------
# TF-IDF MODEL
# -----------------------------------------

print("\nTF-IDF Representation:\n")

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(documents)

# Print feature names
print("Features (Words):")
print(tfidf.get_feature_names_out())

# Convert matrix to array
print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())


# -----------------------------------------
# N-GRAM MODEL
# -----------------------------------------

print("\nBigram Model (N=2):\n")

bigram = CountVectorizer(ngram_range=(2,2))

bigram_matrix = bigram.fit_transform(documents)

print("Bigram Features:")
print(bigram.get_feature_names_out())

print("\nBigram Matrix:")
print(bigram_matrix.toarray())


print("\nTrigram Model (N=3):\n")

trigram = CountVectorizer(ngram_range=(3,3))

trigram_matrix = trigram.fit_transform(documents)

print("Trigram Features:")
print(trigram.get_feature_names_out())

print("\nTrigram Matrix:")
print(trigram_matrix.toarray())
```
