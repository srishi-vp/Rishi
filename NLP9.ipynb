```python
# -*- coding: utf-8 -*-
# EXP NO: 9
# Word Embedding using Word2Vec

# Install required libraries
!pip install beautifulsoup4
!pip install lxml
!pip install gensim

# Import libraries
import nltk
import bs4 as bs
import re
import requests
from gensim.models import Word2Vec
from bs4 import BeautifulSoup

# Download tokenizer
nltk.download('punkt')
nltk.download('punkt_tab') # Added to download the missing resource

# Step 1: Load Wikipedia page
url = "https://en.wikipedia.org/wiki/Global_warming"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)
source = response.text

print("Page loaded successfully")

# Step 2: Parse HTML
soup = BeautifulSoup(source, 'lxml')

text = ""

for paragraph in soup.find_all('p'):
    text += paragraph.text

# Step 3: Text preprocessing
text = re.sub(r'\[[0-9]*\]', ' ', text)
text = re.sub(r'\s+', ' ', text)
text = text.lower()
text = re.sub(r'\d', ' ', text)
text = re.sub(r'\s+', ' ', text)

# Step 4: Tokenization
sentences = nltk.sent_tokenize(text)
sentences = [nltk.word_tokenize(sentence) for sentence in sentences]

print("\nSample Tokenized Sentences:")
print(sentences[:5])

# Step 5: Train Word2Vec model
model = Word2Vec(sentences, vector_size=100, min_count=1)

# Step 6: Vocabulary words
words = model.wv.index_to_key
print("\nVocabulary Words:")
print(words[:20])

# Step 7: Word Vector Example
vector = model.wv['global']
print("\nVector for word 'global':")
print(vector)

# Step 8: Similar Words
similar = model.wv.most_similar('global')

print("\nWords similar to 'global':")
print(similar)
```
